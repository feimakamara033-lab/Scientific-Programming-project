```python
# Top selling products
top_products = data.groupby('ITEMS').agg({
    'QUANTIES': 'sum',
    'TOTAL': 'sum',
    'UNIT_PRICE': 'first'
}).sort_values('TOTAL', ascending=False)

print("Top 10 Products by Revenue:")
print(top_products.head(10))
```

```python
# Average transaction value by product
data['TRANSACTION_VALUE'] = data['TOTAL'] / data['QUANTIES']
product_avg_value = data.groupby('ITEMS')['TRANSACTION_VALUE'].mean().sort_values(ascending=False)
print("\nProducts with Highest Average Transaction Value:")
print(product_avg_value.head(10))
```

```python
# Box plot for sales distribution
plt.figure(figsize=(10, 6))
data.boxplot(column='TOTAL', by='CATEGORY', rot=45)
plt.title('Sales Distribution by Category')
plt.suptitle('')
plt.ylabel('Total Sales (Leones)')
plt.tight_layout()
plt.show()
```

```python
# Total sales by category
category_sales = data.groupby('CATEGORY')['TOTAL'].agg(['sum', 'mean', 'count'])
print("Sales by Category:")
print(category_sales)

# Visualize sales by category
plt.figure(figsize=(10, 6))
category_sales['sum'].sort_values().plot(kind='barh', color='skyblue')
plt.xlabel('Total Sales (Leones)')
plt.title('Total Sales by Product Category')
plt.tight_layout()
plt.show()
```

```python
# Sales by payment method
payment_sales = data.groupby('PAYMENT')['TOTAL'].agg(['sum', 'count'])
print("\nSales by Payment Method:")
print(payment_sales)
```

```python
# Hourly sales analysis
data['HOUR'] = pd.to_datetime(data['TIME'], format='%H:%M').dt.hour
hourly_sales = data.groupby('HOUR')['TOTAL'].sum()

plt.figure(figsize=(10, 6))
hourly_sales.plot(kind='bar', color='coral')
plt.xlabel('Hour of Day')
plt.ylabel('Total Sales (Leones)')
plt.title('Sales by Hour')
plt.xticks(rotation=0)
plt.show()
```

```python
# Create a comprehensive summary
def generate_insights(data):
    insights = {
        'Total Revenue': data['TOTAL'].sum(),
        'Average Transaction': data['TOTAL'].mean(),
        'Total Items Sold': data['QUANTIES'].sum(),
        'Most Popular Category': data['CATEGORY'].mode()[0],
        'Most Used Payment': data['PAYMENT'].mode()[0],
        'Top Cashier': data.groupby('CASHIER')['TOTAL'].sum().idxmax(),
        'Best Selling Item': data.groupby('ITEMS')['QUANTIES'].sum().idxmax(),
        
    }
    return insights

insights = generate_insights(data)
print("="*50)
print("BUSINESS INSIGHTS SUMMARY")
print("="*50)
for key, value in insights.items():
    if 'Revenue' in key or 'Transaction' in key or 'Items sold' in key:
        print(f"{key}: {np.float64(value):,.0f} Leones")
    else:
        print(f"{key}: {value}")
```
