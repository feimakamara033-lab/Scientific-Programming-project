```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
```

```python
data = pd.read_csv(r"C:\Users\HP\Desktop\SuperMarket_Analysis.csv")
data.head()
```

```python
y1_column = data.columns[7]
y2_column = data.columns[8]

mean_value = data[y1_column].mean()
median_value = data[y1_column].median()
mode_value = data[y1_column].mode()[0]
std_value = data[y1_column].std()
var_value = data[y1_column].var()


print("UNIT_PRICE")
print(f"Mean: {mean_value}")
print(f"Median: {median_value}")
print(f"Mode: {mode_value}")
print(f"Standared Deviation: {std_value}")
print(f"Variance: {var_value}")

##################


```

```python
mean_value = data[y2_column].mean()
median_value = data[y2_column].median()
mode_value = data[y2_column].mode()[0]
std_value = data[y2_column].std()
var_value = data[y2_column].var()




print("TOTAL PRICE")
print(f"Mean: {mean_value}")
print(f"Median: {median_value}")
print(f"Mode: {mode_value}")
print(f"Standared Deviation: {std_value}")
print(f"Variance: {var_value}")
```
